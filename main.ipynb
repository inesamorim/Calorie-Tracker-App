```python
class OllamaLLM:
    def __init__(self, model='llama3', host='http://localhost:11434'):
        self.model = model
        self.host = host

    def ask(self, message):
        response = requests.post(
            f"{self.host}/api/generate",
            json={
                "model": self.model,
                "prompt": message,
                "stream": False  # easier to handle non-streaming
            },
            timeout=120
        )
        response.raise_for_status()
        return response.json()

    def summarize(self, message):
        result = self.ask(message)
        return result.get('response', '')
```

# Add to dataset

```python
import numpy as np
import pandas as pd


def add_food(df: pd.DataFrame, alimento: str, **kwargs) -> pd.DataFrame:
    """Add a food to the dataset. Only 'alimento' is required.

    Nutrient values are passed via a dict (columns have spaces/accents):
        df = add_food(df, "Banana", **{"Valor Calórico": 89, "Proteína": 1.1})
    Missing columns are filled with NaN.
    """
    alimento = alimento.strip()

    # Reject duplicates (case-insensitive)
    if df["Alimento"].str.strip().str.lower().eq(alimento.lower()).any():
        raise ValueError(f"'{alimento}' already exists in the dataset.")

    # Reject unknown columns (catches typos)
    unknown = set(kwargs) - set(df.columns)
    if unknown:
        raise KeyError(f"Unknown columns: {sorted(unknown)}")

    row = {col: np.nan for col in df.columns}
    row.update(kwargs)
    row["Alimento"] = alimento

    return pd.concat([df, pd.DataFrame([row])], ignore_index=True)
```

# Save Meal

```python
import json
import os
import pandas as pd

REFEICOES_CSV = "refeicoes.csv"
NAO_SOMAVEIS = ["Alimento", "Densidade Nutricional"]  # identificador e score, não somam


def criar_dataset_refeicoes(food_data_df: pd.DataFrame) -> pd.DataFrame:
    nutrientes = [c for c in food_data_df.columns if c not in NAO_SOMAVEIS]
    return pd.DataFrame(columns=["Nome da refeição", "Alimentos"] + nutrientes)


def carregar_refeicoes(food_data_df: pd.DataFrame, path: str = REFEICOES_CSV) -> pd.DataFrame:
    if os.path.exists(path):
        df = pd.read_csv(path)
        df["Alimentos"] = df["Alimentos"].apply(json.loads)  # str -> lista
        return df
    return criar_dataset_refeicoes(food_data_df)


def add_refeicao(df_ref: pd.DataFrame, food_data_df: pd.DataFrame, nome_refeicao: str,
                 alimentos: list[tuple[str, float]], path: str = REFEICOES_CSV) -> pd.DataFrame:
    """alimentos: lista de (nome, quantidade em gramas).

    df_ref = add_refeicao(df_ref, food_data_df, "Almoço",
                          [("Arroz", 150), ("Frango", 120)])
    """
    nutrientes = [c for c in df_ref.columns if c not in ("Nome da refeição", "Alimentos")]

    base = food_data_df.copy()
    base.index = base["Alimento"].str.strip().str.lower()
    base = base[nutrientes].apply(pd.to_numeric, errors="coerce")

    parcelas = []
    for nome, qtd in alimentos:
        chave = nome.strip().lower()
        if chave not in base.index:
            raise KeyError(f"'{nome}' não existe no food_data_df.")
        parcelas.append(base.loc[chave] * qtd / 100)  # valores por 100 g

    totais = pd.DataFrame(parcelas).sum(min_count=1)  # NaN só se ninguém tiver o valor

    linha = {"Nome da refeição": nome_refeicao.strip(), "Alimentos": alimentos, **totais.to_dict()}
    df_ref = pd.concat([df_ref, pd.DataFrame([linha])], ignore_index=True)

    guardar = df_ref.copy()
    guardar["Alimentos"] = guardar["Alimentos"].apply(json.dumps)  # lista -> str para o CSV
    guardar.to_csv(path, index=False)
    return df_ref
```

# Calculate TMB and calories to consume

To calculate your BMR/RMR using the Mifflin-St. Jeor equationTrusted Source, you’ll need to use one of two equations:

- Males: 10 × weight (in kilograms) + 6.25 × height (in centimeters) – 5 × age (in years) + 5
- Females: 10 × weight (in kilograms) + 6.25 × height (in centimeters) – 5 × age (in years) – 161

```python
def calculate_bmr(weight, height, age, gender):
    """
    Calculate Basal Metabolic Rate (BMR) using the Mifflin-St Jeor Equation.
    
    Parameters:
    weight (float): Weight in kilograms
    height (float): Height in centimeters
    age (int): Age in years
    gender (str): 'male' or 'female'

    Returns:
    float: BMR value
    """

    if gender.lower() == 'male':
        bmr = 10 * weight + 6.25 * height - 5 * age + 5
    elif gender.lower() == 'female':
        bmr = 10 * weight + 6.25 * height - 5 * age - 161
    else:
        raise ValueError("Gender must be 'male' or 'female'")

    return bmr
```

To calculate your TDEE, you’ll need to add an activity factor to the equation:

- 1.2: sedentary (little to no exercise)
- 1.375: lightly active (light exercise 1 to 3 days per week)
- 1.55: moderately active (moderate exercise 6 to 7 days per week)
- 1.725: very active (hard exercise every day, or exercising twice a day)
- 1.9: extra active (very hard exercise, training, or a physical job)

Then, you can calculate your TDEE using this equation:

TDEE = BMR x activity factor

```python
def calculate_tde(bmr, activity_level):
    """
    Calculate Total Daily Energy Expenditure (TDEE) based on BMR and activity level.
    
    Parameters:
    bmr (float): Basal Metabolic Rate
    activity_level (str): One of the following - 'sedentary', 'lightly active', 'moderately active', 'very active', 'extra active'

    Returns:
    float: TDEE value
    """

    activity_multipliers = {
        'sedentary': 1.2,
        'lightly active': 1.375,
        'moderately active': 1.55,
        'very active': 1.725,
        'extra active': 1.9
    }

    if activity_level not in activity_multipliers:
        raise ValueError("Activity level must be one of the following: " + ", ".join(activity_multipliers.keys()))

    tde = bmr * activity_multipliers[activity_level]
    return tde

```

```python
def calorie_deficit(tdee, deficit=400):
    """for weight loss, a common recommendation is to create a calorie deficit of 400-500 calories per day. This function calculates the target daily caloric intake for weight loss based on the provided TDEE."""
    return tdee - deficit
```
